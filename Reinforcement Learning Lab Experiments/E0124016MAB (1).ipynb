```python
import pandas as pd
import random

df = pd.read_csv("Ads_CTR_Optimisation.csv")

n = df.shape[1]
cnt = [0] * n
rew = [0] * n
eps = 0.1

for i in range(len(df)):
    if random.random() < eps:
        ad = random.randint(0, n - 1)
    else:
        avg = []
        for j in range(n):
            if cnt[j] == 0:
                avg.append(0)
            else:
                avg.append(rew[j] / cnt[j])
        ad = avg.index(max(avg))

    r = df.iloc[i, ad]
    cnt[ad] += 1
    rew[ad] += r

ctr = []

for i in range(n):
    if cnt[i] == 0:
        ctr.append(0)
    else:
        ctr.append(rew[i] / cnt[i])

best = ctr.index(max(ctr))

print("Ad Performance:")
for i in range(n):
    print(f"Ad {i}: Shown = {cnt[i]}, Clicks = {rew[i]}, CTR = {ctr[i]:.4f}")

print("\nBest Ad:", best)
print("Best CTR:", max(ctr))
```
