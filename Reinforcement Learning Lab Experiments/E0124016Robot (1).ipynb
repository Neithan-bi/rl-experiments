```python
import numpy as np

states = ["high", "low"]
actions = ["search", "wait", "recharge"]

gamma = 0.9

P = {
    "high": {
        "search": [(0.8, "high", 2), (0.2, "low", 2)],
        "wait": [(1.0, "high", 0)],
        "recharge": [(1.0, "high", 0)]
    },
    "low": {
        "search": [(0.4, "high", 2), (0.6, "low", -3)],
        "wait": [(1.0, "high", 0)],
        "recharge": [(1.0, "high", 0)]
    }
}

V = {s: 0 for s in states}

while True:
    delta = 0

    for s in states:
        old = V[s]
        vals = []

        for a in actions:
            val = 0

            for p, ns, r in P[s][a]:
                val += p * (r + gamma * V[ns])

            vals.append(val)

        V[s] = max(vals)
        delta = max(delta, abs(old - V[s]))

    if delta < 1e-6:
        break

policy = {}

for s in states:
    vals = []

    for a in actions:
        val = 0

        for p, ns, r in P[s][a]:
            val += p * (r + gamma * V[ns])

        vals.append(val)

    policy[s] = actions[np.argmax(vals)]

print("Value Function:")
for s in states:
    print(s, ":", round(V[s], 4))

print("\nOptimal Policy:")
for s in states:
    print(s, ":", policy[s])
```
